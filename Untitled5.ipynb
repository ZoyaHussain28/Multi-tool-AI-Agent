```python
from groq import Groq
import requests
from ddgs import DDGS
import re
```

```python
import warnings

warnings.filterwarnings("ignore", category=DeprecationWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)
```

```python
GROQ_API_KEY = "YOUR_GROQ_API_KEY"
OPENWEATHER_API_KEY = "YOUR_OPENWEATHER_API_KEY"

client = Groq(api_key=GROQ_API_KEY)
```

```python
def get_weather(city):
    try:
        url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid={OPENWEATHER_API_KEY}&units=metric"

        response = requests.get(url).json()

        print(response)  # DEBUG

        if response.get("cod") != 200:
            return f"Couldn't find weather for {city}"

        temp = response["main"]["temp"]
        desc = response["weather"][0]["description"]

        return f"{city}: {temp}°C, {desc}"

    except Exception as e:
        return f"Weather error: {e}"
```

```python
def calculate(expression):
    try:
        result = eval(expression)
        return str(result)
    except Exception:
        return "Invalid calculation"
```

```python
def web_search(query):
    try:
        with DDGS() as ddgs:
            results = list(ddgs.text(query, max_results=3))

        if not results:
            return "No results found"

        answer = ""

        for r in results:
            answer += f"{r['title']}\n"
            answer += f"{r['body']}\n\n"

        return answer

    except Exception as e:
        return f"Search error: {e}"
```

```python
def determine_tool(user_query):

    prompt = f"""
    Determine which tool is needed.

    Available tools:
    - weather
    - calculator
    - search

    User query:
    {user_query}

    Respond with only one word:
    weather, calculator, or search.
    """

    response = ask_groq(prompt)

    return response.strip().lower()
```

```python
def extract_city(user_query):

    prompt = f"""
    Extract only the city name from this weather query.

    Query:
    {user_query}

    Return only the city name.
    """

    response = ask_groq(prompt)

    return response.strip()
```

```python
def ask_groq(prompt):

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content
```

```python
def ai_agent(user_query):

    try:

        tool = determine_tool(user_query)

        if tool == "weather":

            city = extract_city(user_query)

            result = get_weather(city)

        elif tool == "calculator":

            result = calculate(user_query)

        else:

            result = web_search(user_query)

        final_prompt = f"""
        User asked:
        {user_query}

        Tool output:
        {result}

        Give a natural, helpful response.
        """

        return ask_groq(final_prompt)

    except Exception as e:

        return f"Error: {str(e)}"
```

```python
while True:

    q = input("You: ")

    if q.lower() == "exit":
        break

    print("\nAgent:")
    print(ai_agent(q))
    print()
```
